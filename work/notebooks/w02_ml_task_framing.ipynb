## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
For this classification task, the target would be a binary 'yes/no' label representing an observed outcome. For instance, if we're classifying customer churn, the target would be 'churned' (yes) or 'not churned' (no), based on actual customer behavior observed over a future time period. It is crucial that this label comes from an observed phenomenon in the data, not a predefined rule, to ensure the model learns from the real world. A proxy might be used if the direct outcome is hard to measure, but it must still be an observed, measurable event highly correlated with the true outcome.$0
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
For a classification task, a common success metric is ROC-AUC (Receiver Operating Characteristic - Area Under the Curve). Alternatively, precision/recall vs. base rate is also highly relevant, especially in scenarios with imbalanced classes. For example, if predicting customer churn (where churners are a minority), high recall might be prioritized to identify as many churners as possible, even if it means some false positives. The choice depends on the business context and the cost of different types of errors. It's essential to define this metric before model training to ensure an unbiased evaluation of performance.$0
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis represents the single 'row' in your dataset that the model will make a prediction about. For a classification task, this is typically the entity for which you want to predict the binary outcome. For example, if predicting customer churn, the unit of analysis would be a single **customer** at a specific point in time. Each row in the dataframe would represent one customer, with columns containing features (e.g., age, usage, tenure) and the target label (churned/not_churned).

```python
import pandas as pd

# Example of what a unit of analysis dataframe might look like
data = {
    'customer_id': [1, 2, 3, 4, 5],
    'age': [30, 45, 22, 58, 35],
    'tenure_months': [12, 36, 3, 60, 24],
    'monthly_spend': [50.0, 120.5, 25.0, 80.0, 95.0],
    'num_support_calls': [1, 0, 3, 0, 1],
    'churned': [0, 0, 1, 0, 1] # 0 for no churn, 1 for churn (observed outcome)
}
df_unit_of_analysis = pd.DataFrame(data)

display(df_unit_of_analysis.head())
print(f"One row represents one: {df_unit_of_analysis.columns[0].replace('_id', '')}") # Dynamically infer 'customer'
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
Machine Learning beats a fixed rule here because the underlying patterns leading to the target outcome are complex, multi-dimensional, and likely to shift over time. A simple if-statement would struggle to capture the nuanced interactions between various features (e.g., customer age, tenure, support call history, product usage, demographics). It would require manually defining numerous thresholds and rules, which would be brittle, hard to maintain, and quickly become outdated. ML models, however, can learn these intricate relationships from data, adapt to new patterns, and generalize across different scenarios, providing more accurate and robust predictions than a hand-coded set of rules.$0
```
